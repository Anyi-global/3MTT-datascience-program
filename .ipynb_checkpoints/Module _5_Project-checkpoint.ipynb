# Evaluating the Impact of Marketing Campaigns

In a fast paced world like we experience daily, every company seeks one means of the other to maintain profitability and gain quantifiable market share. For marketing companies they focus on consistently running successful campaigns in a complex and competitive market. 

For this to be possible, they set up on A/B testing systems, a randomized experimentation method that involves presenting different versions of a variable (such as a web page, banner, or ad) to separate segments of the audience simultaneously. This approach helps determine which version has the most significant impact on business metrics.

To address these questions, the company conducted an A/B test, where the majority of participants were exposed to ads (the experimental group) and a smaller portion saw a Public Service Announcement (PSA) or nothing (the control group).

## About the Dataset

| **Feature**         | **Description**                                                                                 |
|--------------------|-------------------------------------------------------------------------------------------------|
| **Row Index**      | Row index                                                                                      |
| **User ID**        | Unique identifier for each user                                                                |
| **Test Group**     | Indicates whether the user was in the "ad" group (saw the advertisement) or the "psa" group (saw the public service announcement or nothing) |
| **Converted**      | Indicates whether the user bought the product (True) or not (False)                            |
| **Total Ads**      | The number of ads seen by the user                                                             |
| **Most Ads Day**   | The day when the user saw the highest number of ads                                            |
| **Most Ads Hour**  | The hour of the day when the user saw the highest number of ads                                |

The marketing data can be found **[here](https://drive.google.com/file/d/1YDS_6_1sCJPjucNX6Feo9uGa8B3w8Xzk/view?usp=sharing)**.

## Objective

The primary objective of this case study is to perform hypothesis testing on the e-commerce dataset to compare different groups or features. Specifically, we aim to:

- Analyze the experimental and control groups to determine if the ads were successful.

- Estimate the potential revenue generated from the ads.

- Assess the statistical significance of the differences between the groups using A/B testing techniques.

```python
# Visualize the distribution of total_ads
plt.figure(figsize=(10, 4))
sns.boxplot(x='test_group', y='total_ads', data=df)
plt.title('Total Ads Distribution by Group')

plt.show()
```

**Insights**: <br/>
The boxplots reveal some key insights:

The conversion rate (mean of converted) is very low in both groups, indicating that most users did not convert.
The distribution of total ads shown is quite broad, with a median around 13-14 ads, but some users saw up to 2065 ads, indicating high variability.
Next, to assess whether the differences between the groups are statistically significant, I will perform hypothesis testing. Since the conversion data is binary and the total ads are continuous, I will conduct a t-test for the total ads and a chi-square test for conversion rates.

```python
# Binning total ads into fixed ranges for test group 'ad'
ad_group = df[df['test_group'] == 'ad']
ad_group['ads_bin'] = pd.cut(
    ad_group['total_ads'],
    bins=[0, 10, 20, 50, 100, 500, 1000, ad_group['total_ads'].max()],
    labels=['0-10','11-20','21-50','51-100','101-500','501-1000','1000+'],
    include_lowest=True
)

# Compute conversion rate per bin
conversion_by_ads_bin = ad_group.groupby('ads_bin')['converted'].mean().reset_index()

# Bar plot of conversion rate by total ads seen bin
ax = sns.barplot(data=conversion_by_ads_bin, x='ads_bin', y='converted')
plt.title("Conversion Rate by Ads Seen (Ad Group Only)")
plt.ylabel("Conversion Rate")
plt.xlabel("Ads Seen Bin")

# Add value labels to bars
for i, row in conversion_by_ads_bin.iterrows():
    ax.text(i, row['converted'] + 0.0005, f"{row['converted']:.1%}", ha='center')

plt.show()
```

**Insight**: User groups who saw the most ads were the most to be converted. Users who saw `1000+` ads has the highest conversion rate, meaning 22.2% bought the product after seeing 1000+ ads.

```python
# Binning total ads into fixed ranges for test group 'psa'
psa_group = df[df['test_group'] == 'psa']
psa_group['ads_bin'] = pd.cut(
    psa_group['total_ads'],
    bins=[0, 10, 20, 50, 100, 500, 1000, ad_group['total_ads'].max()],
    labels=['0-10','11-20','21-50','51-100','101-500','501-1000','1000+'],
    include_lowest=True
)

# Compute conversion rate per bin
conversion_by_ads_bin = psa_group.groupby('ads_bin')['converted'].mean().reset_index()

# Bar plot of conversion rate by total ads seen bin
ax = sns.barplot(data=conversion_by_ads_bin, x='ads_bin', y='converted')
plt.title("Conversion Rate by Ads Seen (PSA Group Only)")
plt.ylabel("Conversion Rate")
plt.xlabel("psa Seen Bin")

# Add value labels to bars
for i, row in conversion_by_ads_bin.iterrows():
    ax.text(i, row['converted'] + 0.0005, f"{row['converted']:.1%}", ha='center')

plt.tight_layout()
plt.show()
```

**Insight**: User groups who saw the most psa were the most to be converted. Users who saw `101-500` psa has the highest conversion rate, meaning 11.9% bought the product after seeing those psa.

### Check for Uplift

```python
# Count conversions and totals for both groups
ad_converted = df[df['test_group'] == 'ad']['converted'].sum()
ad_total = df[df['test_group'] == 'ad'].shape[0]

psa_converted = df[df['test_group'] == 'psa']['converted'].sum()
psa_total = df[df['test_group'] == 'psa'].shape[0]
```

```python
# Calculate conversion rates
conversion_rate_ad = ad_converted / ad_total
conversion_rate_psa = psa_converted / psa_total

uplift = conversion_rate_ad - conversion_rate_psa
```

```python
print(f"Ad Group Conversion Rate: {conversion_rate_ad:.3%}")
print(f"PSA Group Conversion Rate: {conversion_rate_psa:.3%}")
print(f"Uplift: {uplift:.3%}")
```

## Hypothesis Testing

**Conversion Rate Hypothesis**: <br>
**Null Hypothesis (H0)**: There is no difference in conversion rates between control and experiment groups.

**Alternative Hypothesis (H1)**: The experiment group has a higher conversion rate.

**z_stat**: how many standard deviations the observed difference is from the expected difference (under the null hypothesis)

**p_val**: If p < 0.05 → the difference is statistically significant, meaning the observed effect is unlikely to be due to random chance

```python
from statsmodels.stats.proportion import proportions_ztest
from statsmodels.stats.proportion import proportion_confint

# Z-test for proportions
ad_count = [ad_converted, psa_converted]
ad_count
```

```python
# Calculate test_group ad_group and psa_tota number of observations
ad_nobs = [ad_total, psa_total]
ad_nobs
```

```python
# Calculate the z-statistic and the p_value to decide whether to Reject or Fail to Reject the Null Hypothesis
z_stat, p_value = proportions_ztest(ad_count, ad_nobs, alternative='larger')

print(f"Z-statistic: {z_stat:.5f}")
print(f"P-value: {p_value:.5f}")
```

**Insight**: The p-value is far below 0.05 signifance level, so we reject the null hypothesis. There is `strong statistical evidence` that the `ad_group had a higher conversion rate than the PSA group`. Therefore, `the uplift is real and statistically significant`.<br><br>

### Estimation

```python
## Attribution & Incremental Conversions
# (0.77% * 564577)

incremental_conversions = uplift * ad_total 
print(f"Estimated Incremental Conversions from Ads: {int(incremental_conversions)}")
```

```python
## Estimation of potential revenue generated from the ads

# Revenue per conversion (assume average amount of product purchased per conversion is $50)
avg_revenue_per_conversion = 50
incremental_revenue = incremental_conversions * avg_revenue_per_conversion

print(f"Estimated Incremental Revenue: ${incremental_revenue:,.2f}")
```

```python
# Summary printout for uplift and statistical significance
print("\n----- Uplift Summary -----")
print(f"Ad Group Conversion Rate: {conversion_rate_ad:.2%}")
print(f"PSA Group Conversion Rate: {conversion_rate_psa:.2%}")
print(f"Absolute Uplift: {uplift:.2%}")
print(f"Z-Statistic: {z_stat:.2f}")
print(f"P-Value: {p_value:.4f}")
print(f"Statistically Significant: {'Yes' if p_value < 0.05 else 'No'}")
print(f"Incremental Conversions: {int(incremental_conversions)}")
print(f"Estimated Incremental Revenue (at ${avg_revenue_per_conversion} per conversion): ${incremental_revenue:,.2f}")
```
