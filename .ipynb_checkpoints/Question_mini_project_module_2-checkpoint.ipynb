## **Data Cleaning (Preparing our Data for Analysis)**

```python
# Check for missing values
missing_values = df[df.isnull().any(axis=1)]
print(missing_values)
```

```python
df.head(20)
```

```python
# count the number of missing values in each row under columns
no_of_missing_values = df.isnull().sum()
no_of_missing_values
```

### **Handle Missing Values**

```python
# Since there are missing values in the 'Product' and 'Total' columns, we will handle them.
# For 'Product', we can fill missing values with 'Unknown Product'.
# For 'Total', we can calculate it based on 'Quantity' and 'Price' if it's missing.

# filling missing values in 'Product'
df['Product'].fillna('Unknown Product')

# Calculating 'Total' where it is missing
missing_total_mask = df['Total'].isnull()
df.loc[missing_total_mask, 'Total'] = df.loc[missing_total_mask, 'Quantity'] * df.loc[missing_total_mask, 'Price']
```

## **Exploratory Data Analysis (EDA)**

### **What's our best performing product by sales?**

```python
# Now that we have handled some missing values, we can proceed with our analysis
# Let's analyze the total sales per product to understand which product is performing best.

# Grouping by 'Product' and summing the 'Total' sales
grouped_prod = df.groupby('product')
product_sales = grouped_prod['total'].sum().reset_index()

# Sorting the results in descending order
product_sales = product_sales.sort_values(by='total', ascending=False)

# Displaying the total sales per product
print(product_sales)
```

From our analysis we can conclude that Laptop is our best performing product with a total sale of 106405.98

### Let's visualize the total sales per product:

```python
# visualizing the total sales per product using a bar chart
plt.figure(figsize=(8, 5))
bars = plt.bar(product_sales['product'], product_sales['total'], color='skyblue')
plt.title('Total Sales per Product')
plt.xlabel('Product')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()

# Add data labels
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # X position
        height,                             # Y position (top of bar)
        f'{height:.2f}',                    # Text (formatted)
        ha='center', va='bottom'            # Align text
    )

# Display the plot
plt.show()
```

### **What's our Total Sales for each Month?**

```python
# Let's calculate the monthly sales for the year 2024

# Extracting the year and month for grouping
# Creating a new column for Year and Month
df['yearmonth'] = df['date'].dt.to_period('M')

# Grouping by Month of the year and summing the 'total' sales
grouped_yearmonth = df.groupby('yearmonth')
monthly_sales = grouped_yearmonth['total'].sum().reset_index()

# Displaying the monthly sales data
print(monthly_sales)
```

```python
monthly_sales.max()
```

```python
monthly_sales.min()
```

### Let's visualize our Monthly Sales for the year

```python
# Visualizing the monthly sales over time using a line chart
plt.figure(figsize=(10, 6))
plt.plot(monthly_sales['yearmonth'].astype(str), monthly_sales['total'], marker='o', color='orange')
plt.title('Monthly Sales Over Time')
plt.xlabel('Year-Month')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()

# Displaying the plot
plt.show()
```

### **What is the Sales Trends by Customer Segment?**

```python
# Extract the year and month for grouping

# Group by customer segment and YearMonth, then sum the total sales
grouped_cust_yearmonth = df.groupby(['customerid', 'yearmonth'])
sales_trends = grouped_cust_yearmonth['total'].sum().reset_index()

# Display the sales trends data
print(sales_trends)
```

### Let's visualize our data

```python
# Pivot the sales trends data for better visualization
pivot_sales = sales_trends.pivot(index='yearmonth', columns='customerid', values='total')

# Plotting the sales trends by customer segment again
plt.figure(figsize=(12, 6))
plt.plot(pivot_sales.index.astype(str), pivot_sales.values, marker='o')
plt.title('Sales Trends by Customer Segment')
plt.xlabel('Year-Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.legend(pivot_sales.columns, title='Customer Segment')
plt.grid()
plt.tight_layout()
plt.show()
```

### **Which Customer Ordered the Most Product?**

```python
# Grouping the data by CustomerID and summing the total orders for each customer
grouped_cust = df.groupby('customerid')
customer_order_totals = grouped_cust['total'].sum().reset_index()

# Sorting the results to find the customer with the highest order value
highest_order_customer = customer_order_totals.sort_values(by='total', ascending=False)

# Displaying the customers with the highest orders in their descending order
print(highest_order_customer)
```

### Let's visualize

```python
# Plotting the total order value for all customers
plt.figure(figsize=(12, 6))
plt.bar(customer_order_totals['customerid'], customer_order_totals['total'], color='lightblue')
plt.title('Total Order Value by Customer')
plt.xlabel('Customer ID')
plt.ylabel('Total Order Value')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```
